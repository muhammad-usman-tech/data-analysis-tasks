```python
from datetime import datetime
# date and time function
now_date = datetime.now()
print(now_date)
# printing the seprate parts of date
print(now_date.year)
print(now_date.day)
print(now_date.date())
print(now_date.month)
# converting a string date into a datetime object
date_str= "2006-05-15"
date_obj = datetime.strptime(date_str,"%Y-%m-%d")
print("Object of the date converted",date_obj)
print(date_obj.month)
date_1=input("Enter 1st date : ")
date_2=input("Enter 2nd date : ")
date1_obj=datetime.strptime(date_1,"%Y-%m-%d")
date2_obj=datetime.strptime(date_2,"%Y-%m-%d")
diff_date=date1_obj-date2_obj
print("The difference between the dates in days are",abs(diff_date.days))
```
