```python
import pandas as pd
import numpy as np
df=pd.read_csv('StudentsPerformance.csv')
df.shape
df.dtypes
# they mentioned marks only we can also sum all the marks first and find the avg but the simple it can be the simple avg of the math score
# student has no name so adding the name
# random data

df['name']=np.random.choice(['ali','usman','ayesha','fatima','nasir','usba'],size=len(df))
df.info()
df.groupby('name')['math score'].mean().sort_values(ascending=True)

highest_math=df['math score'].max()
highest_reading=df['reading score'].max()
highest_writting=df['writing score'].max()
highest_writting
# finding students who failed but in math
df_failed=df[df['math score']<40]
print(len(df_failed)) #40 students failed

# subject wise avg summary does not make sense here because we have different marks but can make the summary by gen
df.groupby('gender')['math score'].mean() # male got more with 68. some thing
df.groupby('gender')['reading score'].mean()  #female got more with 72. some thing
df.groupby('gender')['writing score'].mean()

# identiffing top 5 students by all marks
df['Total_score']=df['math score']+df['reading score']+df['writing score']
df.groupby('name')['Total_score'].sum().sort_values(ascending=False).head() #got all except usman
```
