```python
import pandas as pd
import numpy as np
df=pd.read_csv('/content/melt_rolling_practice_sales.csv')
df.shape

df.info()
df_long=pd.melt(
    id_vars='Month',
    var_name='Store',
    value_name='Sales',
    frame=df
)
df_long

df_long=df_long.sort_values(by='Store',ascending=True)

df_long
month_order = [
    'Jan', 'Feb', 'Mar', 'Apr', 'May', 'Jun',
    'Jul', 'Aug', 'Sep', 'Oct', 'Nov', 'Dec'
]

df_long['Month']=pd.Categorical(
    df_long['Month'],
    categories=month_order,
    ordered=True
)

df_long=df_long.sort_values(by=['Month','Store'],ascending=True)
df_long['rolling avg']=(df_long.groupby('Store')['Sales']
                        .rolling(3)
                        .mean().reset_index(level=0,drop=True))
df_long
```
